```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "google-t5/t5-base"
)
model = AutoModelForSeq2SeqLM.from_pretrained(
    "google-t5/t5-base"
)
```

```python

input_ids = tokenizer("translate English to Deutch: This review examines electrocorrosion processes in reinforced concrete structures caused by stray currents. It evaluates the application of functionally graded cement concrete with controlled electrical conductivity.", return_tensors="pt").to(model.device)

output = model.generate(**input_ids, max_length=512, min_length=100, do_sample=True,top_k=50, temperature=0.7,cache_implementation="static")
print(tokenizer.decode(output[0], skip_special_tokens=True))
```
